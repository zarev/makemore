```python
import os
from PIL import Image
from pdf2image import convert_from_path

def file_to_webp(input_path, output_path):

    # Extract the file name without the extension
    file_name = os.path.splitext(os.path.basename(input_path))[0]

    # Check if the input file is a PDF
    if input_path.lower().endswith('.pdf'):
        # Convert PDF to images
        images = convert_from_path(input_path)
        
        # Save each page as WebP, pdfs can have multiple
        for i, image in enumerate(images):
            image.save(f"/{output_path}/{file_name}_{i+1}.webp", 'WEBP')
        print(f"Converted PDF to WebP: {file_name}_{i+1}.webp")
    
    if input_path.lower().endswith('.png') or input_path.lower().endswith('.jpg'):
        # Open the image file
        with Image.open(input_path) as img:
            # Convert and save as WebP
            img.save(f"/{output_path}/{file_name}.webp", 'WEBP')
        print(f"Converted image to WebP: /{output_path}/{file_name}.webp")
    

def process_file(input_file, output_dir):
    if not os.path.exists(input_file):
        print(f"Error: Input file {input_file} does not exist.")
        return

    output_file = f"/{output_dir}"
    file_to_webp(input_file, output_file)

import os

def process_path(input_path, output_path, process_file_function):
    """
    Processes the input path, either processing all files within a directory
    or processing a single file directly.

    :param input_path: The input path, either a file or a directory
    :param output_path: The output path where processed files should be saved
    :param process_file: A function to process individual files
    """
    if not os.path.exists(input_path):
        print("Error: Input path does not exist.")
        return

    # If the input path is a directory, process all files within it
    if os.path.isdir(input_path):
        for root, _dirs, files in os.walk(input_path):
            for file in files:
                file_path = os.path.join(root, file)
                process_file(file_path, output_path)
    else:
        # If it's a single file, process it directly
        process_file(input_path, output_path)

input_path = '/Users/m/Desktop/gpt/'  # Can be a file or a directory
output_path = 'Users/m/Desktop/gpt/output_webp'

process_path(input_path, output_path, process_file)
```

```python
# Take webp and take out the content. 
from PIL import Image
import pytesseract

def extract_text_from_webp(webp_path):
    # Open the WEBP image file
    image = Image.open(webp_path)

    # Use pytesseract to extract text
    text = pytesseract.image_to_string(image)

    return text

def extract_webp_files(input_directory:str) -> list[dict]:
    """
    Process all WEBP files in a directory, extract text, and return the results.

    :param input_directory: The directory containing WEBP files.
    """
    extracts = []
    # Iterate over all files in the input directory
    for root, _dirs, files in os.walk(input_directory):
        for file in files:
            if file.lower().endswith('.webp'):
                file_path = os.path.join(root, file)
                print(f"Loading file: {file_path}")
                
                # Extract text from the WEBP file and append
                extracted_text = extract_text_from_webp(file_path)
                extracts.append({
                    "File": file,
                    "Extracted text": f"{extracted_text}\n{'=' * 10}"
                })

    return extracts


if __name__ == '__main__':
    webp_path = '/Users/m/Desktop/gpt/output_webp/small_batch/togo_anmeldung_1.webp'  # Path to your WEBP file
    extracted_text = extract_text_from_webp(webp_path)
    print("Extracted Text:\n", extracted_text)
```

```python
import csv

def clean_string(input_string: str, old_string: str, new_string: str) -> str:
    # Replace occurrences of old_string with new_string in the input string
    modified_string = input_string
    if new_string: 
        modified_string = input_string.replace(old_string, new_string)

    # Remove double quotation marks
    modified_string = modified_string.replace('"', '')
    # Remove single quotation marks
    modified_string = modified_string.replace("'", '')

    return modified_string

# Example usage
output_file = 'output.csv'  # Replace with your desired output CSV file path
input_string = 'This is the old string that you want to replace with a new string.'  # The long string to search for
new_string = 'This is the new string that will replace the old string.'  # The new string to replace with

clean_string(input_string, output_file, new_string)

def save_string_to_csv(output_file, input_string):
    # Open the output CSV file in write mode
    with open(output_file, mode='w', newline='', encoding='utf-8') as outfile:
        writer = csv.writer(outfile)
        
        # Split the updated string by commas to create a list
        string_as_list = input_string.split(',')
        
        # Write the list as a single row in the CSV file
        writer.writerow(string_as_list)

# Example usage
updated_string = "This is the new and improved string, that has been successfully updated, with commas as delimiters."
output_file = 'output.csv'  # Replace with your desired output CSV file path

# Save the updated string to a CSV file
save_string_to_csv(output_file, updated_string)

print(f"Updated string saved to {output_file}.")
```
